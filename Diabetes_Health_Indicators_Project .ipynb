```python

from sklearn.feature_selection import mutual_info_classif
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Treat BMI as continuous and the other encoded indicators as discrete
discrete_mask = [col != 'BMI' for col in X_train.columns]

# Calculate feature relevance using training data only
mi_scores = mutual_info_classif(
    X_train,
    y_train,
    discrete_features=discrete_mask,
    random_state=42
)

# Create feature ranking
feature_scores = pd.DataFrame({
    'Feature': X_train.columns,
    'Mutual_Information': mi_scores
})

feature_scores = feature_scores.sort_values(
    by='Mutual_Information',
    ascending=False
)

print("FEATURE IMPORTANCE RANKING")
display(feature_scores)

# Visualize the top 15 features
plt.figure(figsize=(10, 7))

sns.barplot(
    data=feature_scores.head(15),
    x='Mutual_Information',
    y='Feature'
)

plt.title("Top 15 Features Based on Mutual Information")
plt.xlabel("Mutual Information Score")
plt.ylabel("Features")
plt.tight_layout()
plt.show()
```

```python

# Select top 15 features
selected_features = feature_scores.head(15)['Feature'].tolist()

print("Selected Features:")
for feature in selected_features:
    print(feature)

# Create feature-selected training and testing datasets
X_train_selected = X_train[selected_features]
X_test_selected = X_test[selected_features]

print("\nOriginal Training Shape:", X_train.shape)
print("Selected Training Shape:", X_train_selected.shape)

print("\nOriginal Testing Shape:", X_test.shape)
print("Selected Testing Shape:", X_test_selected.shape)
```

```python

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import cross_val_score, GridSearchCV
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Use Age to predict BMI
X_simple_train = X_train[['Age']]
X_simple_test = X_test[['Age']]

y_bmi_train = df.loc[X_train.index, 'BMI']
y_bmi_test = df.loc[X_test.index, 'BMI']

# Train Simple Linear Regression
simple_model = LinearRegression()
simple_model.fit(X_simple_train, y_bmi_train)

# Prediction
simple_predictions = simple_model.predict(X_simple_test)

# Evaluation
print("SIMPLE LINEAR REGRESSION")
print("------------------------")
print("Intercept:", simple_model.intercept_)
print("Coefficient:", simple_model.coef_[0])

print("\nMAE:", mean_absolute_error(y_bmi_test, simple_predictions))
print("MSE:", mean_squared_error(y_bmi_test, simple_predictions))
print("RMSE:", np.sqrt(mean_squared_error(y_bmi_test, simple_predictions)))
print("R2 Score:", r2_score(y_bmi_test, simple_predictions))

# Actual vs predicted graph
plt.figure(figsize=(8, 5))
plt.scatter(y_bmi_test, simple_predictions, alpha=0.15)
plt.xlabel("Actual BMI")
plt.ylabel("Predicted BMI")
plt.title("Simple Linear Regression: Actual vs Predicted BMI")
plt.show()
```

```python

simple_cv_scores = cross_val_score(
    LinearRegression(),
    X_simple_train,
    y_bmi_train,
    cv=5,
    scoring='r2',
    n_jobs=-1
)

print("5-FOLD CROSS-VALIDATION")
for i, score in enumerate(simple_cv_scores, start=1):
    print(f"Fold {i}: {score:.4f}")

print("\nMean R2 Score:", simple_cv_scores.mean())
print("Standard Deviation:", simple_cv_scores.std())
```

```python

simple_grid = GridSearchCV(
    LinearRegression(),
    param_grid={
        'fit_intercept': [True, False],
        'positive': [True, False]
    },
    cv=5,
    scoring='r2',
    n_jobs=-1
)

simple_grid.fit(X_simple_train, y_bmi_train)

print("Best Parameters:", simple_grid.best_params_)
print("Best Cross-Validation R2:", simple_grid.best_score_)

simple_tuned = simple_grid.best_estimator_
simple_tuned_predictions = simple_tuned.predict(X_simple_test)

print("\nTUNED MODEL TEST RESULTS")
print("MAE:", mean_absolute_error(y_bmi_test, simple_tuned_predictions))
print("RMSE:", np.sqrt(mean_squared_error(y_bmi_test, simple_tuned_predictions)))
print("R2:", r2_score(y_bmi_test, simple_tuned_predictions))
```

```python

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)
from sklearn.model_selection import (
    StratifiedKFold,
    cross_validate,
    GridSearchCV
)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create Logistic Regression pipeline
logistic_model = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression(max_iter=1000, random_state=42))
])

# Train the model
logistic_model.fit(X_train, y_train)

# Predict on test data
y_pred_log = logistic_model.predict(X_test)
y_prob_log = logistic_model.predict_proba(X_test)[:, 1]

# Evaluate model
print("LOGISTIC REGRESSION RESULTS")
print("---------------------------")
print("Accuracy:", accuracy_score(y_test, y_pred_log))
print("Precision:", precision_score(y_test, y_pred_log, zero_division=0))
print("Recall:", recall_score(y_test, y_pred_log, zero_division=0))
print("F1 Score:", f1_score(y_test, y_pred_log, zero_division=0))
print("ROC-AUC:", roc_auc_score(y_test, y_prob_log))

print("\nCLASSIFICATION REPORT")
print(classification_report(y_test, y_pred_log, zero_division=0))

# Confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_log,
    display_labels=['No Diabetes', 'Diabetes'],
    cmap='Blues'
)

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Standardize training and testing features
pca_scaler = StandardScaler()

X_train_pca_scaled = pca_scaler.fit_transform(X_train)
X_test_pca_scaled = pca_scaler.transform(X_test)

print("Original Training Shape:", X_train.shape)
print("Original Testing Shape:", X_test.shape)

print("\nScaled Training Shape:", X_train_pca_scaled.shape)
print("Scaled Testing Shape:", X_test_pca_scaled.shape)
```

```python

# Fit PCA on training data
pca_full = PCA()

X_train_pca_full = pca_full.fit_transform(X_train_pca_scaled)
X_test_pca_full = pca_full.transform(X_test_pca_scaled)

# Explained variance
explained_variance = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

pca_results = pd.DataFrame({
    'Principal Component': [
        f'PC{i}' for i in range(1, len(explained_variance) + 1)
    ],
    'Explained Variance': explained_variance,
    'Cumulative Variance': cumulative_variance
})

display(pca_results.round(4))

print("Total Components:", pca_full.n_components_)
```

```python

plt.figure(figsize=(10, 5))

plt.bar(
    range(1, len(explained_variance) + 1),
    explained_variance,
    alpha=0.7,
    label='Individual explained variance'
)

plt.step(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    where='mid',
    label='Cumulative explained variance'
)

plt.axhline(
    y=0.95,
    color='red',
    linestyle='--',
    label='95% variance threshold'
)

plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("PCA - Explained and Cumulative Variance")
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Determine components required for 95% variance
components_95 = np.argmax(cumulative_variance >= 0.95) + 1

print("Original Number of Features:", X_train.shape[1])
print("Components Required for 95% Variance:", components_95)
print(
    "Variance Preserved:",
    round(cumulative_variance[components_95 - 1] * 100, 2),
    "%"
)
```

```python

# Apply PCA with selected number of components
pca_95 = PCA(n_components=0.95)

X_train_pca = pca_95.fit_transform(X_train_pca_scaled)
X_test_pca = pca_95.transform(X_test_pca_scaled)

print("Original Training Shape:", X_train.shape)
print("PCA Training Shape:", X_train_pca.shape)

print("\nOriginal Testing Shape:", X_test.shape)
print("PCA Testing Shape:", X_test_pca.shape)

print(
    "\nTotal Explained Variance:",
    round(np.sum(pca_95.explained_variance_ratio_) * 100, 2),
    "%"
)
```

```python

from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
import pandas as pd

# Hyperparameter combinations
cluster_values = [2, 3, 4, 5]
init_values = ['k-means++', 'random']
n_init_values = [10, 20]

tuning_results = []

for k in cluster_values:
    for init_method in init_values:
        for n_init_value in n_init_values:

            model = KMeans(
                n_clusters=k,
                init=init_method,
                n_init=n_init_value,
                random_state=42
            )

            labels = model.fit_predict(X_kmeans_scaled)

            tuning_results.append({
                'Clusters': k,
                'Initialization': init_method,
                'N_Init': n_init_value,
                'Inertia': model.inertia_,
                'Silhouette': silhouette_score(
                    X_kmeans_scaled, labels
                ),
                'Davies_Bouldin': davies_bouldin_score(
                    X_kmeans_scaled, labels
                ),
                'Calinski_Harabasz': calinski_harabasz_score(
                    X_kmeans_scaled, labels
                )
            })

# Display all configurations
tuning_df = pd.DataFrame(tuning_results)

print("K-MEANS HYPERPARAMETER TUNING RESULTS")
print(tuning_df.round(4).to_string(index=False))

# Select the best configuration using Silhouette Score
best_kmeans_row = tuning_df.loc[
    tuning_df['Silhouette'].idxmax()
]

print("\nBEST K-MEANS CONFIGURATION:")
print(best_kmeans_row)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

# Use the shared 10,000-record test sample
common_test_index = X_knn_test.index
y_common_test = y_knn_test.to_numpy()

# Verify that the other models use the same test records
assert X_dt_test.index.equals(common_test_index)
assert X_svm_test.index.equals(common_test_index)
assert X_bag_test.index.equals(common_test_index)

assert y_dt_test.equals(y_knn_test)
assert y_svm_test.equals(y_knn_test)
assert y_bag_test.equals(y_knn_test)

# Align Logistic Regression predictions to the shared test sample
log_initial_pred = pd.Series(
    y_pred_log, index=X_test.index
).loc[common_test_index].to_numpy()

log_initial_prob = pd.Series(
    y_prob_log, index=X_test.index
).loc[common_test_index].to_numpy()

log_tuned_pred = pd.Series(
    y_pred_log_tuned, index=X_test.index
).loc[common_test_index].to_numpy()

log_tuned_prob = pd.Series(
    y_prob_log_tuned, index=X_test.index
).loc[common_test_index].to_numpy()


# Collect actual predictions from all nine classifiers
model_predictions = [
    ("Logistic Regression",
     log_initial_pred, log_initial_prob,
     log_tuned_pred, log_tuned_prob),

    ("KNN",
     y_knn_pred, y_knn_prob,
     y_knn_tuned_pred, y_knn_tuned_prob),

    ("Decision Tree",
     y_dt_pred, y_dt_prob,
     y_dt_tuned_pred, y_dt_tuned_prob),

    ("Random Forest",
     y_rf_pred, y_rf_prob,
     y_rf_tuned_pred, y_rf_tuned_prob),

    ("SVM",
     y_svm_pred, y_svm_prob,
     y_svm_tuned_pred, y_svm_tuned_prob),

    ("Bagging",
     y_bag_pred, y_bag_prob,
     y_bag_tuned_pred, y_bag_tuned_prob),

    ("AdaBoost",
     y_ada_pred, y_ada_prob,
     y_ada_tuned_pred, y_ada_tuned_prob),

    ("Gradient Boosting",
     y_gb_pred, y_gb_prob,
     y_gb_tuned_pred, y_gb_tuned_prob),

    ("Stacking",
     y_stack_pred, y_stack_prob,
     y_stack_tuned_pred, y_stack_tuned_prob)
]


# Calculate metrics automatically
comparison_rows = []

for model_name, initial_pred, initial_prob, tuned_pred, tuned_prob in model_predictions:

    for stage, predictions, probabilities in [
        ("Initial", initial_pred, initial_prob),
        ("Tuned", tuned_pred, tuned_prob)
    ]:

        comparison_rows.append({
            "Model": model_name,
            "Stage": stage,
            "Accuracy": accuracy_score(y_common_test, predictions),
            "Precision": precision_score(
                y_common_test, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_common_test, predictions, zero_division=0
            ),
            "F1 Score": f1_score(
                y_common_test, predictions, zero_division=0
            ),
            "ROC-AUC": roc_auc_score(
                y_common_test, probabilities
            )
        })


classification_comparison = pd.DataFrame(comparison_rows)

print("CLASSIFICATION MODEL COMPARISON")
print("Common test records:", len(y_common_test))
print("All nine models evaluated on the same test records.")

display(classification_comparison.round(4))
```
